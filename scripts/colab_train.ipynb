Dataset: [Intraoral Tooth Detection](https://universe.roboflow.com/dental-cdueb/intraoral-tooth-detection-rohlq) (CC BY 4.0),

2. Runtime > Change runtime type > **GPU T4**.
3. Jalankan sel 0, 1, 2 berturut-turut (latihan 40-60 menit).
4. Hasil `best.pt` + grafik tersimpan di Drive `smilealign/runs/smilealign-seg/weights/`, unduh ke komputer dan
   taruh di `backend/models/seg/best.pt`.
5. Wajib sitasi: judul dataset + Roboflow 3.0 Instance Seg, lisensi CC BY 4.0 (credit di README).

```python
# Sel 0: mount Drive + ekstrak dataset
from google.colab import drive
drive.mount('/content/drive')

ZIP = '/content/drive/MyDrive/smilealign/intraoral_yolov8.zip'
DATA = '/content/datasets/intraoral'

import zipfile, os
os.makedirs(DATA, exist_ok=True)
with zipfile.ZipFile(ZIP) as z:
    z.extractall(DATA)
print('Splits:', {s: len(os.listdir(f'{DATA}/intraoral/{s}/images')) for s in ['train','valid','test']})
```

```python
# Sel 1: setup ultralytics
!pip -q install ultralytics
import ultralytics
ultralytics.checks()
```

```python
# Sel 2: latihan + simpan ke Drive
from ultralytics import YOLO

model = YOLO('yolov8n-seg.pt')  # pretrained ImageNet/COCO
results = model.train(
    data=f'{DATA}/intraoral/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    project='/content/drive/MyDrive/smilealign/runs',
    name='smilealign-seg',
    patience=15,          # stop awal bila plateau
    lr0=0.01,
    cache=False,
    workers=2,
)
metrics = model.val()  # evaluasi split test/valid
print(metrics.results_dict)
```

Selesai. Unduh dari Drive: `runs/smilealign-seg/weights/best.pt` → taruh di repo lokal
`backend/models/seg/best.pt`.
